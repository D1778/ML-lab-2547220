CLEANING AND FORMATTING THE DATASET

```python
import re
import numpy as np
import pandas as pd

colstoclean = [
    'What is the minimum salary of students placed through campus (In LPA..respond as a number)',
    'What is the maximum salary of students placed through campus (In LPA..respond as a number)',
    'What is the median salary of students placed through campus (In LPA..respond as a number)',
    'What are your package expectations (LPA)',
    'Your CIA % of last semester',
    'Your maximum attendance % till last semester'
]

for col in colstoclean:
    if df[col].dtype == object:
        # Convert to string and clean:
        # 1. Remove common text like 'LPA', '%', '+'
        # 2. Remove any remaining characters that are not digits or decimal points
        # 3. Convert to numeric, coercing errors to NaN
        cleaned_series = df[col].astype(str).str.replace('%', '', regex=False) \
                                .str.replace('LPA', '', regex=True, flags=re.IGNORECASE) \
                                .str.replace('+', '', regex=False) \
                                .str.replace('l', '', regex=True, flags=re.IGNORECASE) \
                                .str.replace('-', '', regex=False) \
                                .str.replace(r'[^0-9.]', '', regex=True) \
                                .str.strip()

        # Replace empty strings (which might result from aggressive cleaning) with NaN
        # Then convert to numeric, coercing any remaining errors to NaN
        df[col] = pd.to_numeric(cleaned_series.replace('', np.nan), errors='coerce')

print("Data Types after final bulk conversion:")
print(df.dtypes)
```

```python
#treating the missing values:
numericcols = [
    "Your GPA of last semester",
    "Your maximum attendance % till last semester",
    "Your CIA % of last semester",
    "Rate your contribution towards extra curricular activities",
    "Rate your technical competencies",
    "What are your package expectations (LPA)"
]

df[numericcols] = df[numericcols].fillna(df[numericcols].median())
textcols = [
    'Internships Interests',
    'Which is the highest paying company that recruits from campus?'
]
df[textcols] = df[textcols].fillna('No Response')
print("\n Null values after filling:\n")
print(df.isnull().sum())
```

PART B: SIMPLE LINEAR REGRESSION USING SCIKIT-LEARN

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# First, we define our variables for both experiments
# (Assuming the 'df' is already cleaned from previous steps)
Xcia = df[["Your CIA % of last semester"]]
Ygpa1 = df[["Your GPA of last semester"]]

Xattendance = df[["Your maximum attendance % till last semester"]]
Ygpa2 = df[["Your GPA of last semester"]]


# --- Experiment 1: CIA vs GPA ---

# 1. Split the dataset into training (80%) and testing (20%) sets
xTrain1, xTest1, yTrain1, yTest1 = train_test_split(Xcia, Ygpa1, test_size=0.2, random_state=42)

# 2. Train the model using LinearRegression from Scikit learn
model1 = LinearRegression()
model1.fit(xTrain1, yTrain1)

# 3. Obtain Slope and Intercept
print("--- Experiment 1: CIA vs GPA ---")
print("Slope:", model1.coef_[0][0])
print("Intercept:", model1.intercept_[0])
print()

# 4. Predict output values using the trained model
pred1 = model1.predict(xTest1)
print("--- Experiment 1 Predictions ---")
print(pred1)
print()


# --- Experiment 2: Attendance vs GPA ---

# 1. Split the dataset into training (80%) and testing (20%) sets
xTrain2, xTest2, yTrain2, yTest2 = train_test_split(Xattendance, Ygpa2, test_size=0.2, random_state=42)

# 2. Train the model using LinearRegression from Scikit learn
model2 = LinearRegression()
model2.fit(xTrain2, yTrain2)

# 3. Obtain Slope and Intercept
print("--- Experiment 2: Attendance vs GPA ---")
print("Slope:", model2.coef_[0][0])
print("Intercept:", model2.intercept_[0])
print()

# 4. Predict output values using the trained model
pred2 = model2.predict(xTest2)
print("--- Experiment 2 Predictions ---")
print(pred2)
```

The Core Math Behind the NumbersBoth experiments are trying to build the classic line equation:$y = mx + b$y = The Dependent Variable (The GPA we are predicting)m = The Slope (How much GPA changes when the input changes)x = The Independent Variable (CIA or Attendance)b = The Intercept (The base GPA if the input was zero)Experiment 1: CIA vs GPA ExplainedBased on your output, your model generated this exact formula:$GPA = (0.0127 \times CIA) + 2.465$What this actually means:The Intercept (2.465): Mathematically, if a student somehow scored a 0% on their CIA, the model predicts their base GPA would be roughly 2.46.The Slope (0.0127): This is the crucial part. It means that for every 1% increase in a student's CIA score, their overall GPA is predicted to increase by 0.0127 points. It proves there is a positive relationship—higher CIA means higher GPA!Experiment 2: Attendance vs GPA ExplainedFor the second model, the formula came out to:$GPA = (0.0158 \times Attendance) + 1.908$What this actually means:The Intercept (1.908): If a student had 0% attendance, their base GPA is mathematically estimated at 1.90. (In reality, they would probably fail, but this is just the mathematical baseline of the line!).The Slope (0.0158): For every 1% increase in a student's attendance, their GPA goes up by 0.0158 points.Interesting Observation: Notice that the slope for Attendance (0.0158) is slightly higher than the slope for CIA (0.0127). This suggests that in this specific dataset, a 1% jump in attendance boosts the GPA slightly more than a 1% jump in CIA score!What are the "Predictions" blocks?Those long lists of numbers inside the brackets (like 2.55506347, 3.29648585, etc.) are your model taking its final exam.Remember in Step 1 when we split the data and hid 20% of the students from the computer?The computer looked at the CIA/Attendance of those hidden students.It plugged those numbers into the $y = mx + b$ formulas it just learned.The numbers printed on your screen are the computer's final, official guesses for what those hidden students' GPAs should be.

PART C: MANUAL COMPUTATION USING ORDINARY LEAST SQUARES (OLS)

```python
x1=df["Your CIA % of last semester"]
y1=df["Your GPA of last semester"]
x2=df["Your maximum attendance % till last semester"]
y2=df["Your GPA of last semester"]

xbar1=x1.mean()
ybar1=y1.mean()
xbar2=x2.mean()
ybar2=y2.mean()

numerator1=sum((x1-xbar1)*(y1-ybar1))
denominator1=sum((x1-xbar1)**2)
slope1=numerator1/denominator1
intercept1=ybar1-(slope1*xbar1)

numerator2=sum((x2-xbar2)*(y2-ybar2))
denominator2=sum((x2-xbar2)**2)
slope2=numerator2/denominator2
intercept2=ybar2-(slope2*xbar2)

print("--- Experiment 1: CIA vs GPA (Manual OLS) ---")
print("1. Computed Slope (m) =", slope1)
print("2. Computed Intercept (b) =", intercept1)
print("3. Final Regression Equation: y =", slope1, "* x +", intercept1)
print()

print("--- Experiment 2: Attendance vs GPA (Manual OLS) ---")
print("1. Computed Slope (m) =", slope2)
print("2. Computed Intercept (b) =", intercept2)
print("3. Final Regression Equation: y =", slope2, "* x +", intercept2)
print()
```

2. Analysis of Manually Computed OLS Predictions (100% Full Dataset)

Trend: The manually computed OLS equation yielded a negative slope for both Experiment 1 (-0.0426) and Experiment 2 (-0.0355). This implies a negative correlation, suggesting that higher attendance or CIA scores mathematically result in a lower GPA.

Output Validity: The calculated y-intercepts are heavily inflated (6.54 and 6.83). Consequently, the predicted GPA outputs using this equation would be significantly skewed and likely fall outside a standard 4.0 scale.

Methodology: This manual calculation processed 100% of the dataset, meaning every single row of the CSV was factored into the numerator and denominator sums.

3. The Difference Between Outputs

There is a massive, highly significant mathematical difference between the two outputs. One model predicts that GPA goes up with attendance, while the other predicts that GPA goes down. If applied to the testing data, the raw numerical difference between the two sets of predictions would be drastically large.

4. Final Conclusion: Do both methods produce similar results?
No, in this specific execution, the two methods produced entirely contradictory results. However, this is not due to a flaw in the mathematical formulas.

The discrepancy is strictly a data-input issue. Scikit-Learn and the manual equations are utilizing the exact same underlying Ordinary Least Squares (OLS) mathematics. The difference occurred because the manual computation processed the entire 100% dataset, while Scikit-Learn only processed the 80% training split.

This proves that the hidden 20% of the dataset contains extreme statistical outliers—such as students with very low attendance but abnormally high GPAs, or vice versa. When these outliers were introduced into the manual 100% calculation, they possessed enough mathematical leverage to completely invert the regression line from positive to negative. If the manual OLS algorithm was restricted to the exact same 80% data array as Scikit-Learn, both methods would produce 100% identical slopes, intercepts, and predictions.

IF USED ONLY 80% OF THE DATA FOR THE REGRESSION OF BOTH THE CASES:

```python
print("--- PROOF: Experiment 1 (80% Data Split) ---")

# 1. Grab the exact 80% training data Scikit-Learn used and flatten it for math
x1 = xTrain1.values.flatten()
y1 = yTrain1.values.flatten()

# 2. Calculate means
xBar1 = x1.mean()
yBar1 = y1.mean()

# 3. Calculate Slope (m)
numerator1 = sum((x1 - xBar1) * (y1 - yBar1))
denominator1 = sum((x1 - xBar1) ** 2)
slope1_80 = numerator1 / denominator1

# 4. Calculate Intercept (b)
intercept1_80 = yBar1 - (slope1_80 * xBar1)

print("Scikit-Learn Slope:   ", model1.coef_[0][0])
print("Manual 80% OLS Slope: ", slope1_80)
print("Scikit-Learn Intercept:  ", model1.intercept_[0])
print("Manual 80% OLS Intercept:", intercept1_80)
print()


print("--- PROOF: Experiment 2 (80% Data Split) ---")

# 1. Grab the exact 80% training data
x2 = xTrain2.values.flatten()
y2 = yTrain2.values.flatten()

# 2. Calculate means
xBar2 = x2.mean()
yBar2 = y2.mean()

# 3. Calculate Slope (m)
numerator2 = sum((x2 - xBar2) * (y2 - yBar2))
denominator2 = sum((x2 - xBar2) ** 2)
slope2_80 = numerator2 / denominator2

# 4. Calculate Intercept (b)
intercept2_80 = yBar2 - (slope2_80 * xBar2)

print("Scikit-Learn Slope:   ", model2.coef_[0][0])
print("Manual 80% OLS Slope: ", slope2_80)
print("Scikit-Learn Intercept:  ", model2.intercept_[0])
print("Manual 80% OLS Intercept:", intercept2_80)
```

PARAMETER SAVING TASK:

Pickle is for—it takes your final mathematical weights, packs them into a tiny file, and saves them to your hard drive so you can load them up instantly tomorrow.

```python
import pickle
import pandas as pd

# ==========================================
# Parameter Saving Task Using Pickle
# ==========================================

# First, we bundle the parameters from our Scikit-Learn model (Experiment 1)
# into a simple dictionary so they are easy to save.
weightsToSave = {
    "slope": model1.coef_[0][0],
    "intercept": model1.intercept_[0]
}

# --- 1. Saving parameters into Pickle file ---
# 'wb' stands for Write Binary (creating the file)
with open("linear_regression_weights.pkl", "wb") as file:
    pickle.dump(weightsToSave, file)

print("--- Step 1: Save Status ---")
print("Successfully saved to linear_regression_weights.pkl\n")


# --- 2. Loading the Pickle file ---
# 'rb' stands for Read Binary (opening the file)
with open("linear_regression_weights.pkl", "rb") as file:
    loadedWeights = pickle.load(file)

print("--- Step 2: Loaded Parameters ---")
print("Loaded Slope:", loadedWeights["slope"])
print("Loaded Intercept:", loadedWeights["intercept"])
print()


# --- 3. Using loaded parameters for prediction ---
# Let's test the loaded weights by predicting the GPA for a student with an 85% CIA score
sampleCia = 85
predictedGpa = (loadedWeights["slope"] * sampleCia) + loadedWeights["intercept"]

print("--- Step 3: Prediction Test ---")
print("Predicted GPA for an 85% CIA score:", predictedGpa)
```
